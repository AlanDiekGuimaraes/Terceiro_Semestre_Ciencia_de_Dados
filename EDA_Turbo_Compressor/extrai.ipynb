```python
import mysql.connector
import pandas as pd

#vw_sao_bernardo

# Função para conectar ao banco de dados MySQL
def conectar_banco():
    conn = mysql.connector.connect(
        host="localhost",           # Substitua pelo seu host
        user="root",                # Substitua pelo seu usuário
        password="525748",       # Substitua pela sua senha
        database="congresso"    # Substitua pelo nome do seu banco
    )
    return conn

# Função para exportar uma tabela para CSV
def exportar_tabela_para_csv(conn, tabela):
    query = f"SELECT * FROM {tabela}"
    df = pd.read_sql(query, conn)
    
    # Salva o DataFrame em um arquivo CSV
    df.to_csv(f"{tabela}.csv", index=False)
    print(f"Tabela {tabela} exportada com sucesso para {tabela}.csv")

# Função principal para exportar todas as tabelas do banco para CSV
def exportar_todas_tabelas(conn):
    cursor = conn.cursor()
    
    # Obtém o nome de todas as tabelas no banco de dados
    cursor.execute("SHOW TABLES")
    tabelas = cursor.fetchall()
    
    # Para cada tabela, exporta para CSV
    for tabela in tabelas:
        tabela_nome = tabela[0]  # Nome da tabela
        exportar_tabela_para_csv(conn, tabela_nome)

# Conectar ao banco de dados
conn = conectar_banco()

# Exportar todas as tabelas para CSV
exportar_todas_tabelas(conn)

# Fechar a conexão
conn.close()
```

```python
import pandas as pd

# Carregar os arquivos CSV
dados_turbo = pd.read_csv('Dados_TurboCompressor_KAIK.csv')
vw_sao_bernardo = pd.read_csv('vw_sao_bernardo.csv')

# Verificar tipos de dados antes de concatenar
print("Tipos de dados de 'Dados_TurboCompressor_KAIK.csv':")
print(dados_turbo.dtypes)

print("\nTipos de dados de 'vw_sao_bernardo.csv':")
print(vw_sao_bernardo.dtypes)

# Forçar a conversão das colunas de 'timestamp' para datetime para garantir consistência
dados_turbo['timestamp'] = pd.to_datetime(dados_turbo['timestamp'], errors='coerce')
vw_sao_bernardo['timestamp'] = pd.to_datetime(vw_sao_bernardo['timestamp'], errors='coerce')

# Verificar se houve algum erro na conversão das datas
print("\nVerificação de 'timestamp' após conversão:")
print(dados_turbo['timestamp'].head())
print(vw_sao_bernardo['timestamp'].head())

# Verificar valores nulos
print("\nValores nulos em 'Dados_TurboCompressor_KAIK.csv':")
print(dados_turbo.isnull().sum())

print("\nValores nulos em 'vw_sao_bernardo.csv':")
print(vw_sao_bernardo.isnull().sum())

# Tratar valores nulos (substituir por 0 ou outro valor conforme necessário)
dados_turbo.fillna(0, inplace=True)
vw_sao_bernardo.fillna(0, inplace=True)

# Alinhar os tipos de dados das colunas para garantir que as concatenações sejam consistentes
for col in dados_turbo.columns:
    if col in vw_sao_bernardo.columns:
        vw_sao_bernardo[col] = vw_sao_bernardo[col].astype(dados_turbo[col].dtype)

# Verificar novamente tipos de dados após conversão
print("\nTipos de dados após conversão para garantir consistência:")
print(vw_sao_bernardo.dtypes)

# Concatenar os dois DataFrames
dados_combinados = pd.concat([dados_turbo, vw_sao_bernardo], ignore_index=True)

# Verificar se os dados foram concatenados corretamente
print("\nPrimeiras linhas do DataFrame combinado:")
print(dados_combinados.head())
```
