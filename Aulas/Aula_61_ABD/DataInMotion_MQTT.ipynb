```python
import pandas as pd
import paho.mqtt.client as mqtt
import matplotlib.pyplot as plt
import json
from datetime import datetime

# Configurações MQTT
broker = "test.mosquitto.org"
porta = 1883
topico = "aula/data_in_motion/temperatura_umidade"

# Cria DataFrame vazio
dados = pd.DataFrame(columns=["timestamp", "temperatura", "umidade"])

# Função de callback quando uma mensagem é recebida
def on_message(client, userdata, msg):
    global dados

    try:
        # Decodifica a mensagem JSON
        mensagem = json.loads(msg.payload.decode())
        print(f"📥 Dados recebidos: {mensagem}")

        # Converte para DataFrame temporário (usando concat em vez de append)
        novo_dado = pd.DataFrame([{
            "timestamp": datetime.fromisoformat(mensagem["timestamp"]),
            "temperatura": mensagem["temperatura"],
            "umidade": mensagem["umidade"]
        }])

        # Adiciona ao DataFrame principal (forma moderna)
        dados = pd.concat([dados, novo_dado], ignore_index=True)

        # Atualiza o gráfico
        atualizar_grafico()

    except Exception as e:
        print(f"❌ Erro ao processar mensagem: {e}")

# Função para atualizar o gráfico
def atualizar_grafico():
    plt.clf()  # Limpa o gráfico anterior

    # Plota temperatura e umidade
    plt.plot(dados["timestamp"], dados["temperatura"], label="Temperatura (°C)", color="red")
    plt.plot(dados["timestamp"], dados["umidade"], label="Umidade (%)", color="blue")

    plt.xlabel("Tempo")
    plt.ylabel("Valores")
    plt.title("Monitoramento de Temperatura e Umidade")
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.pause(0.1)  # Pausa breve para atualizar o gráfico

# Configura o cliente MQTT (usando a versão mais recente da API)
cliente = mqtt.Client(mqtt.CallbackAPIVersion.VERSION2)  # Usando a versão 2 da API
cliente.on_message = on_message

# Conecta ao broker
cliente.connect(broker, porta)
print("🟢 Conectado ao broker MQTT.")

# Assina o tópico
cliente.subscribe(topico)
print(f"🔔 Assinando o tópico: {topico}")

# Configura o modo interativo para o matplotlib
plt.ion()

try:
    # Inicia o loop MQTT
    cliente.loop_forever()
except KeyboardInterrupt:
    print("\n🛑 Programa encerrado pelo usuário")
finally:
    cliente.disconnect()
    plt.ioff()
    plt.show()
```
