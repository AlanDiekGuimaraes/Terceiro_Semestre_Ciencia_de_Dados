```python
clientes = [
    {"id": 1, "nome": "João Silva", "idade": 28, "email": "joao.silva@email.com"},
    {"id": 2, "nome": "Maria Oliveira", "idade": 34, "email": "maria.oliveira@email.com"},
    {"id": 3, "nome": "Carlos Souza", "idade": 22, "email": "carlos.souza@email.com"}
]
```

```python
import json

# Salvar os dados em um arquivo
with open("clientes.json", "w") as arquivo:
    json.dump(clientes, arquivo, indent=4)

print("Dados armazenados com sucesso em 'clientes.json'.")
```

```python
# Ler os dados do arquivo
with open("clientes.json", "r") as arquivo:
    dados_carregados = json.load(arquivo)

# Visualizar os dados carregados
print("Dados carregados do arquivo:")
for cliente in dados_carregados:
    print(cliente)
```

```python
# Atualizar o email de Maria
for cliente in dados_carregados:
    if cliente["id"] == 2:
        cliente["email"] = "maria.novo@email.com"

# Salvar novamente
with open("clientes.json", "w") as arquivo:
    json.dump(dados_carregados, arquivo, indent=4)

print("Dados atualizados com sucesso.")
```

```python
# Remover cliente com id 3
dados_carregados = [cliente for cliente in dados_carregados if cliente["id"] != 3]

# Salvar novamente o novo conjunto de dados
with open("clientes.json", "w") as arquivo:
    json.dump(dados_carregados, arquivo, indent=4)

print("Cliente removido com sucesso.")
```
