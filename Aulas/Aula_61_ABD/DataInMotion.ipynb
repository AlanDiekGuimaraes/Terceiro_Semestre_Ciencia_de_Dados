```python
import time
import random
import pandas as pd

# Função para gerar uma nova leitura de sensor
def gerar_leitura_sensor():
    temperatura = round(random.uniform(20.0, 30.0), 2)  # temperatura entre 20 e 30 graus
    umidade = round(random.uniform(30.0, 70.0), 2)       # umidade entre 30% e 70%
    return {"temperatura": temperatura, "umidade": umidade, "timestamp": pd.Timestamp.now()}

# Simulando um fluxo de dados
for i in range(10):  # 10 leituras, uma a cada segundo
    dados = gerar_leitura_sensor()
    print(dados)
    time.sleep(1)
```

```python
# Coletar dados durante o fluxo
leituras = []

for i in range(10):
    leitura = gerar_leitura_sensor()
    leituras.append(leitura)
    print(f"Nova leitura: {leitura}")
    time.sleep(1)

# Criar um DataFrame com os dados recebidos
df = pd.DataFrame(leituras)

# Visualizar todas as leituras
print("\nTodas as Leituras Coletadas:")
print(df)

# Verificar condições em tempo real
for index, row in df.iterrows():
    if row["temperatura"] > 28.0:
        print(f"⚠️ Alerta! Temperatura alta detectada: {row['temperatura']}°C às {row['timestamp']}")
```

```python
import matplotlib.pyplot as plt

# Plotar Temperatura e Umidade ao longo do tempo
plt.figure(figsize=(10,5))
plt.plot(df["timestamp"], df["temperatura"], label="Temperatura (°C)", marker="o")
plt.plot(df["timestamp"], df["umidade"], label="Umidade (%)", marker="x")
plt.title("Fluxo de Dados de Sensor: Temperatura e Umidade")
plt.xlabel("Tempo")
plt.ylabel("Leitura")
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
