```python
import pandas as pd

# Gerando manualmente um Small Data
dados_clientes = {
    "Nome": ["Ana", "Bruno", "Carlos", "Diana", "Eduardo"],
    "Idade": [28, 34, 22, 45, 31],
    "Cidade": ["São Paulo", "Rio de Janeiro", "Curitiba", "Belo Horizonte", "Porto Alegre"],
    "Compras Realizadas": [5, 2, 1, 7, 3]
}

# Transformando em DataFrame
df = pd.DataFrame(dados_clientes)

# Visualizando o pequeno conjunto de dados
print(df)
```

```python
# Média de idade
media_idade = df["Idade"].mean()

# Total de compras realizadas
total_compras = df["Compras Realizadas"].sum()

print(f"Média de idade dos clientes: {media_idade:.1f} anos")
print(f"Total de compras realizadas: {total_compras} compras")
```

```python
import matplotlib.pyplot as plt

# Gráfico de barras simples
plt.bar(df["Nome"], df["Compras Realizadas"], color='lightgreen')
plt.title("Compras por Cliente")
plt.xlabel("Cliente")
plt.ylabel("Número de Compras")
plt.grid(axis='y')
plt.show()
```

```python
# Estatísticas rápidas
print(df.describe())
```

```python
# Relação entre idade e número de compras
plt.scatter(df["Idade"], df["Compras Realizadas"], color='coral')
plt.title("Idade vs Compras Realizadas")
plt.xlabel("Idade")
plt.ylabel("Compras Realizadas")
plt.grid()
plt.show()
```
